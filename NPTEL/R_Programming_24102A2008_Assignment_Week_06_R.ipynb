Loops – for loop

```r
for(i in 1:5) {
    print(i^2)
}
```

```r
for(i in c(2, 3, 4, 5)) {
    print(i^2)
}
```

```r
x = c(2, 4, 6, 8, 10, 12)
```

```r
excount = function(x) {
  count = 0

  for(xval in x) {
    if(xval/2 > 3) {
      count = count+1
    }
    print(count)
  }
}

excount(x)
```

Nested looping with  for loop


```r
child = c("c1", "c2", "c3")
sweet = c("s1", "s2", "s3")
```

```r
for(x in child) {
  for(y in sweet) {
    print(paste(x, y))
  }
}
```

The   break command


```r
drink = c("coffee", "lemonade", "tea", "juice")
```

```r
for(x in drink) {
  if(x == "tea") {
    break
  }
  print(x)
}
```

The   next command


```r
drink = c("coffee", "lemonade", "tea", "juice")
```

```r
for(x in drink) {
  if(x == "lemonade") {
    next
  }
  print(x)
}
```

The   while loop

```r
i = 1
while(i<10) {
  print(i)
  i = i+1
}
```

```r
sumfunction = function() {
  sum = 0
  number = as.integer(readline(prompt="Please select any number less than 25: "))

  while(number <= 25) {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers: ", sum))
}
```

```r
sumfunction()
```

The   repeat loop

```r
i = 1
repeat{
  print(i)
  i = i+2
  if(i > 10) break
}
```

```r
i = 1
repeat{
  i = i+2
  if(i < 10) next
  print(i)
  if(i > 20) break
}
```
